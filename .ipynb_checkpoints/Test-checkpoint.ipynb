```python
import numpy as np
import h5py

with h5py.File(FILE_NAME, 'r') as f:
        print(f"Successfully opened HDF5 file: '{FILE_NAME}'")
        print("\n--- Contents of the file ---")

        # Iterate through top-level items (groups and datasets)
        if len(f.keys()) == 0:
            print("The file is empty or contains no top-level datasets/groups.")

        def print_hdf5_item(name, obj, indent=0):
            prefix = "  " * indent
            if isinstance(obj, h5py.Group):
                print(f"{prefix}📦 **Group:** {name}/")
                for key, item in obj.items():
                    print_hdf5_item(key, item, indent + 1)
            elif isinstance(obj, h5py.Dataset):
                print(f"{prefix}📊 **Dataset:** {name}")
                print(f"{prefix}    Shape: {obj.shape}")
                print(f"{prefix}    Dtype: {obj.dtype}")
                # Optionally, print a slice of data if it's small enough
                if obj.size < 20 and obj.ndim < 2: # Keep output minimal
                    print(f"{prefix}    Data (first few): {obj[()]}")


        f.visititems(print_hdf5_item) # Recursively visit all items in the file

        print("\n--- Example: Accessing a specific dataset after discovery ---")
        # Let's say you discovered 'sensor_readings' and want to read it
        if 'scrambled' in f:
            scrambled_data = f['scrambled'][()]
            print(f"Successfully read 'scrambled' dataset. First 5 rows:\n{scrambled_data[:5]}")
        else:
            print("Dataset 'scrambled' not found (this part might not run if you change the dummy data).")

```
